```python
import time
from datetime import datetime
import json
import random
```

```python
def generate_patientIds():
    pids = [id for id in range(2000, 2050)]
    return pids    
```

```python
def generate_readings(pids):    
    count = 0
    #for iterate in range(0,5):
    while True:
        data = {}
        readings = []
        for x in range(0,10):
            for y in range(0, 50):
                data['patient_id'] = pids[y]
                data['bp_systolic'] = random.randint(100,160)
                data['bp_diastolic'] = random.randint(50, 100)
                data['heart_rate'] = random.randint(60,120)
                data['temperature'] = random.uniform(97.0, 103.5)
                data['spo2'] = random.randint(85,100)
                data['event_time'] = str(datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
                readings.append(data.copy())   
            time.sleep(1)
        fname = 'test'+str(count)+'.json'
        files = open('./data/'+fname, 'w')
        files.write(json.dumps(readings))
        files.close()
        count += 1           
    
```

```python
def capture_data():
    pids = generate_patientIds()
    generate_readings(pids)
    
```

```python
capture_data()
```
